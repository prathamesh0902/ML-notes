## Graphs

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import lasso_path

# -----------------------------
# Generate sample data
# -----------------------------
X, y, coef = make_regression(
    n_samples=100,
    n_features=6,
    n_informative=5,
    noise=10,
    coef=True,
    random_state=42
)

# Standardize features
X = StandardScaler().fit_transform(X)

# -----------------------------
# Compute Lasso coefficient path
# -----------------------------
alphas, coefs, _ = lasso_path(X, y)

# -----------------------------
# Plot coefficient paths
# -----------------------------
plt.figure(figsize=(10,6))

for i in range(coefs.shape[0]):
    plt.plot(alphas, coefs[i], label=f'Feature {i+1}')

plt.xscale('log')
plt.gca() #.invert_xaxis()      # Small alpha on left, large alpha on right

plt.xlabel('Regularization Parameter (Alpha)')
plt.ylabel('Coefficient Value')
plt.title('Lasso Regression: Coefficient Paths')

plt.axhline(0, color='black', linestyle='--')
plt.grid(True)
plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))

plt.tight_layout()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

# Generate data
X, y = make_regression(
    n_samples=200,
    n_features=8,
    n_informative=5,
    noise=15,
    random_state=42
)

# Standardize features
X = StandardScaler().fit_transform(X)

# Different alpha values
alphas = np.logspace(-3, 2, 60)

coef_history = []

# Train Ridge for each alpha
for alpha in alphas:
    model = Ridge(alpha=alpha)
    model.fit(X, y)
    coef_history.append(model.coef_)

coef_history = np.array(coef_history)

# Plot
plt.figure(figsize=(10,6))

for i in range(coef_history.shape[1]):
    plt.plot(alphas, coef_history[:, i], label=f'Feature {i+1}')

plt.xscale('log')
plt.xlabel("Alpha (Regularization Strength)")
plt.ylabel("Coefficient")
plt.title("Ridge Regression: Coefficient Shrinkage")

plt.axhline(0, color='black', linestyle='--')
plt.grid(True)
plt.legend(bbox_to_anchor=(1.02,1), loc='upper left')

plt.tight_layout()
plt.show()
```

## Coefficients

```python
import numpy as np
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler

# Simulated data
np.random.seed(42)
X = np.random.randn(100, 4)
y = 2*X[:,0] + 3*X[:,1] + 40*X[:,2] + 500*X[:,3] + np.random.randn(100)*0.1

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply Lasso
alphas = [0.001, 0.01, 0.1, 1, 10, 100]  # λ values
print(f"{'Alpha':>8} | {'Coef x1':>8} {'Coef x2':>8} {'Coef x3':>8} {'Coef x4':>8}")
print("-" * 50)

for alpha in alphas:
    lasso = Lasso(alpha=alpha, max_iter=10000)
    lasso.fit(X_scaled, y)
    coefs = lasso.coef_
    print(f"{alpha:8.3f} | {coefs[0]:8.3f} {coefs[1]:8.3f} {coefs[2]:8.3f} {coefs[3]:8.3f}")
```

```python
import numpy as np
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler

# 1. Simulate data
np.random.seed(42)
n_samples = 100
X = np.random.randn(n_samples, 4)

# True relationship (OLS-like)
y = 2*X[:, 0] + 3*X[:, 1] + 40*X[:, 2] + 500*X[:, 3] + np.random.randn(n_samples) * 0.1

# 2. Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Try various lambda (alpha) values
alphas = [0.001, 0.01, 0.1, 1, 10, 100]  # λ values
print(f"{'Alpha':>8} | {'Coef x1':>8} {'Coef x2':>8} {'Coef x3':>8} {'Coef x4':>8}")
print("-" * 50)

for alpha in alphas:
    ridge = Ridge(alpha=alpha, max_iter=10000)
    ridge.fit(X_scaled, y)
    coefs = ridge.coef_
    print(f"{alpha:8.3f} | {coefs[0]:8.3f} {coefs[1]:8.3f} {coefs[2]:8.3f} {coefs[3]:8.3f}")
```
